# Further Preprocessing for Power BI

```python
import pandas as pd
```

## Load Data

```python
patterns = pd.read_pickle('preprocessed_patterns.pkl')
```

```python
patterns = patterns.rename(columns = {'has_uk_terminology': 'uk_terminology', 'has_us_terminology': 'us_terminology'})
```

## Add Columns

```python
# Create unknown_terminology column
patterns['unknown_terminology'] = ((patterns['craft'] == 'crochet') & (patterns['uk_terminology'] == False) & (patterns['us_terminology'] == False)).astype(bool)
```

```python
# Create 2-yr sales & revenue columns
patterns['est_revenue_2yr'] = patterns['projects_per_day'] * patterns['price_usd'] * 730
patterns['pattern_sales_2yr'] = patterns['projects_per_day'] * 730
```

## Save

```python
patterns.to_csv('capstone_ravelry.csv', index = False)
```
